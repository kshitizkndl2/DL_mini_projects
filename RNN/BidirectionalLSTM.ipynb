```python
import numpy as np
import tensorflow
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Bidirectional, Dense,LSTM,GRU
from sklearn.metrics import accuracy_score
```

```python
(x_train,y_train),(x_test,y_test) = imdb.load_data(
    num_words=10000,
)
```

```python
len(x_train)
```

```python
len(y_test)
```

```python
x_train = pad_sequences(x_train, maxlen = 200, padding='post', truncating = 'post')
```

```python
x_test = pad_sequences(x_test, maxlen = 200, padding='post', truncating = 'post')
```

```python
model = Sequential([
    Embedding(input_dim=10000, output_dim=32, input_shape=(200,)),
    Bidirectional(LSTM(5, return_sequences=True)),
    Bidirectional(LSTM(5, return_sequences=True)),
    Bidirectional(LSTM(5, return_sequences=True)),
    Bidirectional(LSTM(5)),  # Last LSTM returns only the final sequence output (2D)
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam',loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
stopping = tensorflow.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=1,
    verbose=1,
    mode="auto",

    restore_best_weights=True,
    start_from_epoch=0,
)
```

```python
history = model.fit(x_train,y_train,validation_split = 0.2,epochs = 100,callbacks=[stopping])
```

```python
y_pred = model.predict(x_test)
```

```python
y_a = []
for i in y_pred:
  if i > 0.5:
    y_a.append(1)
  else:
    y_a.append(0)

```

```python
accuracy = accuracy_score(y_test,y_a)
print(accuracy)
```
