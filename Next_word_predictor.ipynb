```python
practice_text = """The sun rises in the east and sets in the west. Every morning, the birds begin to sing as the sky turns bright. The warm sun warms the earth and gives energy to all living things. When evening comes, the sun slowly disappears below the horizon, and the stars appear in the dark sky. This daily cycle of the sun and stars has continued for billions of years, bringing light and darkness in a steady, predictable rhythm. During the day, people work under the bright sky, and during the night, they rest under the quiet stars. The cycle of day and night governs all life on the planet. Plants grow when the sun shines, and animals rest when the darkness falls. Year after year, season after season, this eternal dance of light and shadow continues without end.

In the spring, warm winds blow and beautiful flowers bloom under the gentle sunlight. The fields turn green, and new life begins to wake up from its long winter sleep. Farmers go out to the fields to plant seeds, hoping for a rich harvest. In the summer, the days are long and hot, filled with bright sunshine from morning until late evening. Children play outside under the clear blue sky, and rivers flow with cool water. In the autumn, the leaves turn gold, orange, and red, falling to the ground as the air grows cool and crisp. The harvest is gathered, and everyone prepares for the cold days ahead. In the winter, cold snow covers the silent earth like a white blanket, and the nights become long and dark. The wind is chilly, but families gather warmly inside beside their cozy fires.

Through every change of the season, the sun keeps rising in the east and setting in the west, guiding the world through time. People have learned to adapt to these changes, building houses to stay warm in the winter and finding shade to stay cool in the summer. They watch the stars at night to navigate their journeys across the land and sea. Time never stops, and the natural world continues its beautiful, endless loop. This rhythm of life is shared by every creature on earth, connected by the simple, powerful journey of the sun."""

print(f"Text loaded. Total character count: {len(practice_text)}")
```

```python
import tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([practice_text])
```

```python
tokenizer.word_index
```

```python
import re
sentences = re.split(r'(?<=\.)\s*', practice_text)
print(sentences)
```

```python
tokens = [tokenizer.texts_to_sequences([sentence])[0]for sentence in sentences]
```

```python
print(tokens)
```

```python
input_lst = []
output_lst = []

for seq in tokens:
    for j in range(len(seq) - 1):
        input_lst.append(seq[:j+1])
        output_lst.append(seq[j+1])
```

```python
input_lst[9]
```

```python
output_lst[9]
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
max_len = max([len(x) for x in input_lst])
```

```python
padded_input = pad_sequences(input_lst, maxlen = max_len, padding = 'pre')
```

```python
c = 0
for x in input_lst:


  if len(x) == 24:
    print(c)
  c = c +1

```

```python
padded_input
```

```python
x = padded_input
y = output_lst
```

```python
x.shape
```

```python
import numpy as np
y = np.array(y)
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y,num_classes=198)
```

```python
y.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Embedding,LSTM
```

```python
model = Sequential()
model.add(Embedding(198,100,input_shape = (24,)))
model.add(LSTM(150))
model.add(Dense(198,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
```

```python
model.summary()
```

```python
x
```

```python
history = model.fit(x, y, epochs = 100)
```

```python
text = 'the fields turn'
token_text = tokenizer.texts_to_sequences([text])[0]
```

```python
padded_text = pad_sequences([token_text],maxlen = max_len, padding = 'pre')
```

```python
pos = np.argmax(model.predict(padded_text))
```

```python
for word, index in tokenizer.word_index.items():
  if index == pos:
    print(word)
```
